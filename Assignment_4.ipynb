# **Dictionary Methods Practice:**

```python
# Dictionary Methods Practice

# 1. Create a dictionary called student with key-value pairs
student = {
    "name": "Haroon Rasheed",
    "age": 20,
    "grade": "A",
    "courses": ["Math", "Science", "History"],
}
print(student)

# Print the student's name
print(student["name"])

# Print the list of courses
print(student["courses"])

# Add a new key-value pair: "email"
student["email"] = "haroon.rasheed@example.com"
print(student)


# 2. Modify Dictionary Values
student = {
    "name": "Haroon Rasheed",
    "age": 20,
    "grade": "A",
    "courses": ["Math", "Science", "History"],
}
print(student)

# Update age, courses and grade
student["age"] = 21
student["courses"] = "Computer Science"
student["grade"] = "A+"
print(student)


# 3. Dictionary Operations
student_address = {
    "street": "123 College Ave",
    "city": "New York",
    "zip": "10001"
}

# Combine both dictionaries into a new dictionary called student_info
student_info = {**student, **student_address}

# Remove the "zip" key from student_info
student_info.pop("zip", None)

# Check if "phone" exists in the dictionary
if "phone" in student_info:
    print("phone number exists.")
else:
    print("phone number does not exist.")

# Print the final dictionary
print(student_info)


# 4. Dictionary Methods
word_count = {
    "hello": 5,
    "world": 10,
    "python": 15
}

# Get all keys and print them as a list
print(list(word_count.keys()))

# Get all values and print them as a list
print(list(word_count.values()))

# Get all key-value pairs and print them
print(list(word_count.items()))

# Make a copy of the dictionary
word_count_copy = word_count.copy()

# Clear the original dictionary
word_count.clear()

# Print both dictionaries to verify one is empty and one is a copy
print("Original Dictionary (after clear):", word_count)
print("Copied Dictionary:", word_count_copy)
```

# **Shopping Discount Calculator**

```python
# Set up the initial variables:
print("Welcome to the Discount Calculator📠!")
print("----------------------------------")
print("Welcome to Muhammad Hamza Projects❤❤")
print("----------------------------------")

purchase_amount = 120.00
is_loyalty_member = True
day_of_week = "Saturday"
is_holiday_season = True

developer_name = "Muhammad Hamza"

# Store original purchase amount
original_amount = purchase_amount
final_price = purchase_amount

# Print original amount
print(f"Original purchase amount: ${original_amount:.2f}")

# Apply discount rules using if statements:
# If purchase amount is greater than $100, apply a 10% discount
if purchase_amount > 100:
    discount = final_price * 0.10
    final_price -= discount
    print(f"Purchase Discount (10%): -${discount:.2f}")

# If purchase amount is greater than 50 but less or equal to 100, apply 5% discount
elif purchase_amount > 50:
    discount = final_price * 0.05
    final_price -= discount
    print(f"Purchase Discount (5%): -${discount:.2f}")

# If the customer is a loyalty member, apply an additional 2% discount
if is_loyalty_member:
    discount = final_price * 0.02
    final_price -= discount
    print(f"Loyalty Discount (2%): -${discount:.2f}")

# If it's a weekend (Saturday or Sunday), apply an additional 3% discount
if day_of_week == "Saturday" or day_of_week == "Sunday":
    discount = final_price * 0.03
    final_price -= discount
    print(f"{day_of_week} Discount (3%): -${discount:.2f}")

# If it's a holiday season, apply an additional 8% discount
if is_holiday_season:
    discount = final_price * 0.08
    final_price -= discount
    print(f"Holiday Discount (8%): -${discount:.2f}")

# Final price
print(f"\nFinal Price after all discounts: ${final_price:.2f}")

print("Thank you for using the Discount Calculator📠!")
print("----------------------------------")
print("Thank you for using Muhammad Hamza Projects❤❤")
print("----------------------------------")

print(f"Developed by: {developer_name}")
```

# **While Loop Assignment - Interactive Game**

```python
import random

# 🎲 1. Set a secret number between 1 and 50
secret_num = random.randint(1, 50)
guess_count = 0

print("🔐 Welcome to the Secret Number Game!")
print("🤔 I'm thinking of a number between 1 and 50.")
print("🎯 Can you guess what it is?\n")

# Developer info
developer_name = "Muhammad Hamza"

# 2. Start the guessing loop
while True:
    try:
        guess_number = int(input("🔢 Enter your guess: "))
        guess_count += 1

        # 3. Check the guess
        if guess_number == secret_num:
            print(f"\n🎉 Congratulations! You guessed the number {secret_num} in {guess_count} tries! 🥳")
            break
        elif guess_number < secret_num:
            print("⬇️ Too low! Try again. 💡")
        else:
            print("⬆️ Too high! Try again. 🔍")

    except:
        print("⚠️ Please enter a valid number! 🧠")

    finally:
        print("Thank you for using the Number Guessing Game!")

print(f"\nDeveloped by: {developer_name}")
7
```

# **Student Grade Management System**

```python
students_data = [
    {
        "name": "Haroon",
        "id": 12,
        "grades": {
            "web designing": 80,
            "AI": 90,
            "Graphic": 60,
        }
    },
    {
        "name": "Rasheed",
        "id": 14,
        "grades": {
            "web designing": 90,
            "AI": 60,
            "Graphic": 70,
        }
    },
    {
        "name": "Hamza",
        "id": 18,
        "grades": {
            "web designing": 70,
            "AI": 50,
            "Graphic": 40,
        }
    },
    {
        "name": "Taha",
        "id": 20,
        "grades": {
            "web designing": 85,
            "AI": 75,
            "Graphic": 65,
        }
    }
]

new_student = {
    "name": "Ali",
    "id": 21,
    "grades": {
        "web designing": 88,
        "AI": 76,
        "Graphic": 91,
    }
}
students_data.append(new_student)

print("📋 All Students with their Grades:\n")
for student in students_data:
    print(f"Name: {student['name']}, ID: {student['id']}, Grades: {student['grades']}")

print("\n📊 Average Grade for Each Student:")
for student in students_data:
    avg = sum(student["grades"].values()) / len(student["grades"])
    print(f"{student['name']} - Average Grade: {avg:.2f}")

top_student = max(
    students_data,
    key=lambda student: sum(student["grades"].values()) / len(student["grades"])
)
highest_avg = sum(top_student["grades"].values()) / len(top_student["grades"])
print(f"\n🏅 Student with the Highest Average Grade:\n{top_student['name']} - Average Grade: {highest_avg:.2f}")

print("\n🎖️ Students with Average Grade Above 80:")
for student in students_data:
    avg = sum(student["grades"].values()) / len(student["grades"])
    if avg > 80:
        print(f"{student['name']} - Average Grade: {avg:.2f}")

course_name = "AI"
total = 0
count = 0
for student in students_data:
    if course_name in student["grades"]:
        total += student["grades"][course_name]
        count += 1

if count > 0:
    course_avg = total / count
    print(f"\n📘 Average Grade for Course '{course_name}': {course_avg:.2f}")
else:
    print(f"\nCourse '{course_name}' not found for any student.")
```
